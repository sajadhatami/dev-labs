```python
from abc import ABC, abstractmethod


class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass



class CashPayment(Payment):

    def pay(self, amount):
        print(f"Cash payment: {amount}")



class CardPayment(Payment):

    def pay(self, amount):
        print(f"Card payment: {amount}")



class FreePayment:

    def pay(self):
        print("No payment required")
```
