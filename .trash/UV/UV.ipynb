```python
# init project:

uv init {dir:optional} {--flags}

# flags: --app & --lib
# app is default, if no flags are provided, it will create an app project
# lib is for creating a library project
```

```python
# install packages
uv add {package_name}
# or
uv add -r requirements.txt
```

```python
# activating venv by uv :
uv venv
source .venv/bin/activate
```

```python
# run the project
uv run {order}
```

```python
# for sync envirement:
uv sync
```

```python
# for removing dependencies:
uv remove {package_name}
```

```python
# clone project that have uv:
git clone https://github.com/username/project-name.git
cd project-name
uv sync
```

```python
# install permanant packages
uv tool install {ruff}

# uninstall permanant packages
uv tool uninstall {ruff}

# list of tools:
uv tool list

# update tools:
uv tool upgrade {name or flag: --all}
```
