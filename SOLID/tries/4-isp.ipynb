```python
# ref:
from abc import ABC, abstractmethod


class Worker(ABC):

    @abstractmethod
    def work(self):
        pass

class Eater(ABC):
    @abstractmethod
    def eat(self):
        pass



class HumanWorker(Worker, Eater):

    def work(self):
        print("Working")


    def eat(self):
        print("Eating")



class RobotWorker(Worker):

    def work(self):
        print("Working")
```

```python
# dirty
from abc import ABC, abstractmethod


class FileManager(ABC):

    @abstractmethod
    def read(self, path):
        pass

    @abstractmethod
    def write(self, path, data):
        pass

    @abstractmethod
    def delete(self, path):
        pass

    @abstractmethod
    def compress(self, path):
        pass

    @abstractmethod
    def encrypt(self, path):
        pass



class LocalFileManager(FileManager):

    def read(self, path):
        print(f"Reading {path}")


    def write(self, path, data):
        print(f"Writing {data} to {path}")


    def delete(self, path):
        print(f"Deleting {path}")


    def compress(self, path):
        print(f"Compressing {path}")


    def encrypt(self, path):
        print(f"Encrypting {path}")



class CloudStorage(FileManager):

    def read(self, path):
        print(f"Reading {path} from cloud")


    def write(self, path, data):
        print(f"Uploading {data}")


    def delete(self, path):
        print(f"Deleting {path}")


    def compress(self, path):
        raise Exception("CloudStorage cannot compress files")


    def encrypt(self, path):
        raise Exception("CloudStorage cannot encrypt files")
```

```python
# refactored:
from abc import ABC, abstractmethod


class FileReader(ABC):

    @abstractmethod
    def read(self, path):
        pass

class FileEditor(ABC):
    @abstractmethod
    def write(self, path, data):
        pass
    @abstractmethod
    def delete(self, path):
        pass
    
class FileCompressor(ABC):
    @abstractmethod
    def compress(self, path):
        pass

class FileEncryptor(ABC):
    @abstractmethod
    def encrypt(self, path):
        pass



class LocalFileManager(FileReader, FileEditor, FileCompressor, FileEncryptor):

    def read(self, path):
        print(f"Reading {path}")


    def write(self, path, data):
        print(f"Writing {data} to {path}")


    def delete(self, path):
        print(f"Deleting {path}")


    def compress(self, path):
        print(f"Compressing {path}")


    def encrypt(self, path):
        print(f"Encrypting {path}")



class CloudStorage(FileReader, FileEditor):

    def read(self, path):
        print(f"Reading {path} from cloud")


    def write(self, path, data):
        print(f"Uploading {data}")


    def delete(self, path):
        print(f"Deleting {path}")
```
