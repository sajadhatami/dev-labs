```python
class OrderService:

    def __init__(self):
        self.database = MySQLDatabase()
        self.payment = StripePayment()
        self.email = EmailSender()


    def create_order(self, user, products, payment_type):

        # Check inventory
        for product in products:
            if product.stock <= 0:
                raise Exception("Product unavailable")


        # Calculate price
        total_price = 0

        for product in products:
            total_price += product.price


        # Discount
        if user.type == "vip":
            total_price *= 0.8

        elif user.type == "employee":
            total_price *= 0.5


        # Payment
        if payment_type == "card":
            self.payment.pay(total_price)

        elif payment_type == "cash":
            print("Cash payment")


        # Save order
        self.database.save({
            "user": user,
            "products": products,
            "price": total_price
        })


        # Send notification
        self.email.send(
            user.email,
            "Your order created"
        )


        # Generate invoice
        invoice = Invoice()
        invoice.create(
            user,
            products,
            total_price
        )


class MySQLDatabase:

    def save(self, data):
        print("Saving to MySQL")


class StripePayment:

    def pay(self, amount):
        print(f"Paying {amount} with Stripe")


class EmailSender:

    def send(self, email, message):
        print("Sending email")


class Invoice:

    def create(self, user, products, price):
        print("Creating invoice")
```
